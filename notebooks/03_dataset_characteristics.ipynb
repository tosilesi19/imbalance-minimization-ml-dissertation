# 03: Dataset Characteristics Table

Generates the dataset characteristics summary (output #2 of the study's
output inventory), destined for Chapter Four, Participants and Research
Setting, as Table 2. Reads both processed files so reported figures are
derived, not typed. Expected: Adult 32,561 x 42; Diabetes 101,766 x 108.

```python
import pandas as pd
import numpy as np

BUCKET = "osilesi-dissertation-data-2026"   # your bucket name
pd.set_option("display.max_columns", None)
```

```python
rows = []
for name, path in [("Adult (Financial)", "adult_final_pruned.csv"),
                   ("Diabetes (Healthcare)", "health_final_pruned.csv")]:
    df = pd.read_csv(f"s3://{BUCKET}/processed/{path}")
    rows.append({
        "Dataset": name,
        "Observations": df.shape[0],
        "Features": df.shape[1] - 1,
        "Positive class %": round(df["target"].mean() * 100, 1),
        "Female %": round(df["sex_female"].mean() * 100, 1),
        "Missing values": int(df.isna().sum().sum())
    })

chars = pd.DataFrame(rows)
chars.to_csv("dataset_characteristics.csv", index=False)
chars
```
